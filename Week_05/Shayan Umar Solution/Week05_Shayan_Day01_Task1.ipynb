**----------------------------------------Building an ANN from Scratch-------------------------------------------**

Importing necessary libraries

```python
import numpy as np
import random
```

Initializing Weights and Biases

```python
def initialize_parameters(input_size, hidden_size, output_size):
    np.random.seed(42)  # For reproducibility

    parameters = {
        "W1": np.random.randn(hidden_size, input_size) * 0.01,
        "b1": np.zeros((hidden_size, 1)),
        "W2": np.random.randn(output_size, hidden_size) * 0.01,
        "b2": np.zeros((output_size, 1))
    } # this dictionary will store the values of weights and biases and will be used further in the code
    return parameters
```

Activation Functions

```python
def sigmoid(Z):
    return 1 / (1 + np.exp(-Z))

def relu(Z):
    return np.maximum(0, Z)

def relu_derivative(Z):
    return (Z > 0).astype(int)
```

Forward Propagation

```python
def forward_propagation(X, parameters):
    W1, b1, W2, b2 = parameters["W1"], parameters["b1"], parameters["W2"], parameters["b2"] # get the weights and biases from the parameters dictionary

    Z1 = np.dot(W1, X) + b1 # output of first layer(only hidden layer in our case)
    A1 = relu(Z1)
    Z2 = np.dot(W2, A1) + b2 # final output and our prediction

    A2 = sigmoid(Z2) # squishes any input value in between 0 and 1 but sum is not equal to 1 as it is in softmax

    cache = {"Z1": Z1, "A1": A1, "Z2": Z2, "A2": A2}
    return A2, cache
```

Computing Cost function

```python
def compute_cost(Y, A2):
    m = Y.shape[1] # m is the number of examples .shape[1] means the number of columns in the matrix.
    cost = -np.sum(Y * np.log(A2) + (1 - Y) * np.log(1 - A2)) / m # binary cross entropy loss function
    return np.squeeze(cost) # removes extra dimensions so the cost is returned as a scalar instead of an array like (1,).
```

Backward Propagation

```python
# the overall purpose of this function is to minimize loss/cost function by changing the values of weights and biases.

def backward_propagation(X, Y, parameters, cache):
    m = X.shape[1]
    W2 = parameters["W2"]

    dZ2 = cache["A2"] - Y
    dW2 = np.dot(dZ2, cache["A1"].T) / m
    db2 = np.sum(dZ2, axis=1, keepdims=True) / m

    dZ1 = np.dot(W2.T, dZ2) * relu_derivative(cache["Z1"])
    dW1 = np.dot(dZ1, X.T) / m
    db1 = np.sum(dZ1, axis=1, keepdims=True) / m

    grads = {"dW1": dW1, "db1": db1, "dW2": dW2, "db2": db2}
    return grads

```

Updating parameters to minimize cost function

```python
# actual gradient descent

def update_parameters(parameters, grads, learning_rate):
    for key in parameters.keys():
        parameters[key] -= learning_rate * grads["d" + key]
    return parameters


```

Training Neural Network

```python
# training the neural network by using all the above functions

def train_neural_network(X, Y, input_size, hidden_size, output_size, epochs=1000, learning_rate=0.01):
    parameters = initialize_parameters(input_size, hidden_size, output_size)

    for i in range(epochs):
        A2, cache = forward_propagation(X, parameters)
        cost = compute_cost(Y, A2)
        grads = backward_propagation(X, Y, parameters, cache)
        parameters = update_parameters(parameters, grads, learning_rate)

        if i % 100 == 0:
            print(f"Epoch {i}: Cost = {cost}")

    return parameters


```

Prediction

```python
# prediction using forward propagation

def predict(X, parameters):
    A2, _ = forward_propagation(X, parameters)
    return (A2 > 0.5).astype(int)

```

Training the model and performing prediction


```python
# Example data (AND logic gate)
# X = np.array([[0, 0, 1, 1], [0, 1, 0, 1]])
# Y = np.array([[0, 0, 0, 1]])

X = np.array([[0, 0, 1, 1], [0, 1, 0, 1]])
Y = np.array([[0, 0, 0, 1]])

Y_or = np.array([[0, 1, 1, 1]])

Y_xor = np.array([[0, 1, 1, 0]])



trained_parameters = train_neural_network(X, Y_xor, input_size=2, hidden_size=4, output_size=1, epochs=2000, learning_rate=0.1)

predictions = predict(X, trained_parameters)
print("Predictions:", predictions)


```
